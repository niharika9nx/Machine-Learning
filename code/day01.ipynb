```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age":        [22, 25, 30, 35, 40],
    "Salary":     [30000, np.nan, 50000, np.nan, 75000],
    "Experience": [1, 2, np.nan, 8, 12],
    "City":       ["Delhi", np.nan, "Mumbai", "Pune", np.nan],
    "Gender":     ["M", "F", np.nan, "F", "M"],
    "Department": ["IT", "HR", "Finance", np.nan, "IT"]
})

print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.info())
```

```python
print(df.head(10))
print(df.tail(10))
```

```python
print(df.isnull().sum())
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df.isnull().sum())
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
# city is a categorical value
print(df["City"].isnull().sum())
```

```python
print("Before:", len(df))

df = df.drop_duplicates()

print("After:", len(df))
```

```python
X = df[["Age", "Salary", "Experience"]]

print(X.head())
```

```python
high_salary = df[df["Salary"] > 50000]

print(high_salary)
```

```python
df_sorted = df.sort_values(
    by="Salary",
    ascending=False
)

print(df_sorted.head(10))
```

```python
df_sorted = df.sort_values(
    by="Salary",
    ascending=True
)

print(df_sorted.head(10))
```

```python
# feature engineering
df["Salary_per_Experience"] = (
    df["Salary"] / (df["Experience"] + 1)
)
print(df.head())
```

```python
df["Gender_Code"] = df["Gender"].astype("category").cat.codes
print(df[["Gender", "Gender_Code"]].head())
```

```python
df = pd.get_dummies(
    df,
    columns=["City"],
    dtype=int
)

print(df.head())
```

```python
# this is one one the important feature of pandas - groupby
summary = df.groupby("Department")["Salary"].agg(
    ["mean", "min", "max", "count"]
)

print(summary)
```

```python

Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]

print(outliers)
```
