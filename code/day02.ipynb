```python
import numpy as np
import pandas as pd

df = pd.read_csv("ctc.csv")
```

```python
df['Previous CTC'] = df['Previous CTC'].astype(str).str.replace(',','').astype(float)
df['CTC'] = df['CTC'].astype(str).str.replace(',','').astype(float)

print("Previous CTC dtype: ",df["Previous CTC"].dtype)
print("CTC dtype: ",df['CTC'].dtype)
```

```python
print("College Distribution: ")
print(df["College"].value_counts())
print("\nRole distribution:")
print(df["Role"].value_counts())
print("\nCity type:")
print(df["City type"].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plb

plb.figure(figsize=(15,5))

plb.subplot(1,3,1)
sns.countplot(x="College",data=df,palette='viridis')
plb.title("College Tier Distribution")

plb.subplot(1,3,2)
sns.countplot(x='Role', data=df, palette='viridis')
plb.title("Role Distribution")

plb.subplot(1,3,3)
sns.countplot(x="City type", data=df, palette='magma')
plb.title("City Type Distribution")


plb.tight_layout()
plb.show()

```

```python
# pie plot

import matplotlib.pyplot as plb
plb.figure(figsize=(15,5))
# College tier distrubution

plb.subplot(1,3,1)
college_counts = df['College'].value_counts()
plb.pie(college_counts.values, labels=college_counts.index, autopct='%1.1f%%', startangle=90)
plb.title('College tier distribution')

#Role Distribution 
plb.subplot(1,3,2)
role_counts = df['Role'].value_counts()
plb.pie(role_counts.values, labels=role_counts.index, autopct='%1.1f%%', startangle=90)
plb.title('Role Distribution')

# City Type Distribution
plb.subplot(1,3,3)
city_counts = df['City type'].value_counts()
plb.pie(city_counts.values, labels=city_counts.index, autopct='%1.1f%%',startangle=90)
plb.title('City type Distribution')

plb.tight_layout()
plb.show()

```

```python
# Hist plot

# target distribution
plb.figure(figsize=(12,5))

# Target distribution
plb.subplot(1,2,1)
sns.histplot(df['CTC'], kde=True, color='teal', bins=30)
plb.title("CTC Histplot Distribution")

# box plot

plb.subplot(1,2,2)
sns.boxplot(x= df['CTC'], color='skyblue')
plb.title('CTC Boxplot Distribution')

plb.tight_layout()
plb.show()

print('CTC Skewness:', round(df['CTC'].skew(),3))

```

```python
# Hist plot

# target distribution
plb.figure(figsize=(12,5))

# Target distribution
plb.subplot(1,2,1)
sns.histplot(df['Previous CTC'], kde=True, color='teal', bins=30)
plb.title("Previous CTC Histplot Distribution")

# box plot

plb.subplot(1,2,2)
sns.boxplot(x= df['Previous CTC'], color='skyblue')
plb.title('Previous CTC Boxplot Distribution')

plb.tight_layout()
plb.show()

print('CTC Skewness:', round(df['CTC'].skew(),3))
```

```python
plb.figure(figsize=(15,5))

plb.subplot(1,3,1)
sns.boxplot(x='College', y='CTC', data=df, palette='hls')
plb.title('CTC vs College tier')

plb.subplot(1,3,2)
sns.boxplot(x='Role',y='CTC',data=df, palette='viridis')
plb.title('ctc VS Role')

plb.subplot(1,3,3)
sns.boxplot(x='City type', y='CTC',data=df, palette='magma')
plb.title('CTC vs City type')

plb.tight_layout()
plb.show()
```

```python
plb.figure(figsize=(15, 5))

plb.subplot(1, 3, 1)
sns.scatterplot(x='Previous CTC', y='CTC', data=df, alpha=0.6)
plb.title('CTC vs Previous CTC')

plb.subplot(1, 3, 2)
sns.scatterplot(x='Exp (Months)', y='CTC', data=df, alpha=0.6, color='green')
plb.title('CTC vs Experience (Months)')

plb.subplot(1, 3, 3)
sns.scatterplot(x='Graduation marks', y='CTC', data=df, alpha=0.6, color='orange')
plb.title('CTC vs Graduation Marks')

plb.tight_layout()
plb.show()
```
